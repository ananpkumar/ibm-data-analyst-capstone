# Data Collection: Programming Language Salaries (Web Scraping)

Part of my IBM Data Analyst Capstone project. In this notebook I scrape the **programming language** and **average annual salary** from a web page, using `requests` and `BeautifulSoup`, and save the results to `popular-languages.csv`.

The page is a static table provided with the course, so the figures are illustrative rather than a live salary survey.

```python
from bs4 import BeautifulSoup
import requests
import csv
```

## 1. Download the web page

```python
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"

data = requests.get(url).text
soup = BeautifulSoup(data, "html.parser")
```

## 2. Extract the language name and average annual salary
Find the table, skip the header row, and read the language (column 2) and salary (column 4) from each remaining row.

```python
table = soup.find("table")

rows_to_save = []
for row in table.find_all("tr")[1:]:   # [1:] skips the header row
    cols = row.find_all("td")
    if len(cols) >= 4:
        language = cols[1].getText().strip()
        salary = cols[3].getText().strip()
        rows_to_save.append([language, salary])
        print(language, "--->", salary)
```

## 3. Save to CSV

```python
with open("popular-languages.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Language", "Average Annual Salary"])
    writer.writerows(rows_to_save)
```
