## 5. Average age of respondents
Age is recorded as ranges (for example "25-34 years old"), so I mapped each range to an approximate midpoint to calculate the mean. "Under 18" is set to 17 and "65 years or older" to 70. The result is therefore an estimate, not an exact average.

```python
age_mapping = {
    "Under 18 years old": 17,
    "18-24 years old": 21,
    "25-34 years old": 29,
    "35-44 years old": 39,
    "45-54 years old": 49,
    "55-64 years old": 59,
    "65 years or older": 70
}

# Create a new numeric Age column
df['Age_numeric'] = df['Age'].map(age_mapping)
```

```python
# Calculate the mean age
mean_age = df['Age_numeric'].mean()
print("Mean age of survey participants:", mean_age)
```
